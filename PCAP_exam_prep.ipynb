# PCAP Exam prep

- A namespace is a space in which names exist
- If you want to import *pi* from *math* you will use: **from math import pi**
- Packages can contain modules
- When a module entity should be trated as private 
    - You can mark the entity name with the __ (double underscore) prefix
    - You can mark the entity name with the _ (single underscore) prefix
- A PWG-lead repository, collecting open-source Python code is called **PyPI**
- PyPI is often reffered to as Cheese Shop
- **pip** 
    - It's a command-line tool
    - The name pip comes from *pip installs packages*
    - There are two different pip implementations, one for Python 2 and the other for Python 3
    - When you use pip to install a package that requires one or more dependencies pip will take care of everything by itself
    - You can determine pip version by using following commands:
        - *pip version*
        - *pip -version*
        - *pip --version*
