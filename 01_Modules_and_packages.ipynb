## Math module

```python
import math
```

```python
for name in dir(math):
    print(name, end='\t')
```

```python
# PCEP must know functions
# ceil(), floor(), trunc()
print(math.ceil(3.6))
print(math.floor(3.6))
print(math.trunc(3.6))
```

```python
# ceil () - always rounds the number upwards to the nearest integer
print(math.ceil(6.2))
print(math.ceil(0.0001))
print(math.ceil(9.00000000000001))
print(math.ceil(-8.8)) # up to -8
```

```python
# floor() - opposite to ceil, always downwards
print(math.floor(6.9999))
print(math.floor(0.0001))
print(math.floor(9.00000000000001))
print(math.floor(-8.8)) # down to -9
```

```python
# trunc() - removes decimal part, returns integer from the number
print(math.trunc(6.9999))
print(math.trunc(0.0001))
print(math.trunc(9.00000000000001))
print(math.trunc(-8.8)) # keeps only -8
```

```python
# factorial()
# 3! = 3 * 2 * 1 = 6
# 4! = 4 * 3 * 2 * 1 = 24
```

```python
print(math.factorial(4))
```

```python
# sqrt() - returns the square root of a given number, gives back float
print(math.sqrt(16)) # 4 * 4 = 16
print(math.sqrt(100)) # 10*10 = 100
```

```python
# hypot() --> hypotenuse = c^2 = a^2 + b^2
print(math.hypot(3,4)) # 3^3 + 4*4= 9 + 16 = 25 = c^2
```
