```python
import sys
def exit():
    print('I wanna exit')
exit() # my exit function
sys.exit() # sys exit function
```

```python
# when need to use the exit function defaultly from sys:
from sys import exit
exit()
```

```python
# my exit function will overwrite imported exit function
from sys import exit

def exit():
    print('I wanna exit')
exit() # my exit function
```

```python
from sys import *   # not recommended
```

```python
import sys  # you need to add sys.
from sys import * # no need to add sys., but may get a lot of conflicting names
```

```python
# aliases
from sys import exit as bye_bye
bye_bye()
```
