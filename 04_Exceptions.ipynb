## Propagating Exceptions

```python
def get_day(user_info):
    day = int(input('What is the day of your bday?'))
    user_info.append(day)

def get_month(user_info):
    month = int(input('hat is the month (1-12) of your bday? '))
    user_info.append(month)

def get_year (user_info):
    year = int(input( 'What is the year of your bday? '))
    user_info.append(year)

def get_user_bday(user_info): # function calling 3 previous functions
    try:
        get_day(user_info)
        get_month(user_info)
        get_year(user_info)
        print('So your bday is', user_info)
    except ValueError:
        print('You entered incorrect data, sorry!')

print('Hi, I will collect some info about your bday!')
user_info = []
get_user_bday(user_info)

# RULE: exceptions are propagated through functions in Python
```

## Assertion Exceptions

- assertions = assumptions in our program that should always be true
    - used for debugging, testing code and for documenting 
    - not used for validating user input or handle errors

```python
def calculate_inverse(number):
    assert (number != 0), 'Got 0 as number!'
    return 1/number

calculate_inverse(0)
```

## The finally keyword
- must be the very last branch of the code intended for handling exceptions
- even after the "else" block
- is executed always no matter what

```python
def get_inverse(x):
    try:
        return 1/x
    except ZeroDivisionError:
        return None
    finally:
        print('I am always printed')
```

```python
print(get_inverse(0))
```

## The raise keyword

```python
def return_bigger(a, b):
    if not isinstance(a, int) or not isinstance(b, int):
        raise ValueError
    if b>a:
        return b
    else:
        return a
```

```python
print(return_bigger(5,3))
```

```python
def return_reverse(x):
    try:
        return 1/x
    except:
        print('something went wrong') # prints output
        raise # raises error
```

```python
return_reverse(0) # output printed + ZeroDivisionError raised
```

```python
# What will be the output?
class MyEx(Exception):
  def __init__(self, msg):
    Exception.__init__(self, msg+msg)
    self.args = (msg,)
 
try:
  raise MyEx('wrong!')
except Exception as e:
  print(e)
```
