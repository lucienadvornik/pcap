## Searching inside strings

```python
# index() - returns the index of the first occurrence of the given value
'ilovetravellingaroundtheworld'.index('a')
# if cannot find - gives back error
# can be used with various types of sequences like strings or lists
```

```python
# find() - returns the index of the first occurrence of the given value, safer to use
'ilovetravellingaroundtheworld'.find('x')
# if cannot find - gives back minus one
# only works with strings
```

```python
# find() with 2 and 3 argument verions
my_text = ' This is my sample string'
my_text.find('is', 10) # starting at index 10 incl until the end of the string
my_text.find('is', 10, 15) # starting at index 10 incl until 15 excl
```

```python
# rfind() - short for right find - starts from the end of the string
print('work work work'.find('work'))
print('work work work'.rfind('work'))
print('work work work'.find('work', 2))
print('work work work'.rfind('work', 2))
print('work work work'.find('work', 2, 5))
print('work work work'.rfind('work', 2, 5))
```

```python
# isalnum() - method that checks if the string contains only alphanumeric characters
print('Adrian'.isalnum())
print('Adrian30'.isalnum())
print('Adrian_30'.isalnum())
print('Adrian 30'.isalnum())
```

```python
# isalpha() - checks letters only
print('Adrian'.isalpha())
```

```python
#isdigit() - checks digits only
print('Adrian'.isdigit())
```

```python
# islower() - checks lowercase
# isupper() - checks highercase
# isspace() - checks white spaces only, new line character also accepted
```

## Joining, splitting and sorting strings

```python
# join()
' '.join(['This is', 'a spectacular', 'place to be'])
```

```python
# split()
'How may\nstrings\nwill you see?'.split()
```

```python
# sorted() - creates new list
names = ['Adam', 'Kate', 'Eva', 'Tom']
names_sorted = sorted(['Adam', 'Kate', 'Eva', 'Tom'])
print(names_sorted)
```

```python
# sort() - changes original list
names = ['Adam', 'Kate', 'Eva', 'Tom']
names.sort()
print(names)
```
