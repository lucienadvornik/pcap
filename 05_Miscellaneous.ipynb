# map() and filter() with lambdas

map()
- takes two arguments, function and sequence with some elements
- map() then applies the function to all the elements in the sequence

```python
lambda_func = lambda i: i*2
initial_list = [1, 2, 3, 4, 5]
```

```python
map_result = map(lambda_func, initial_list)
print(next(map_result))
print(next(map_result))
print(next(map_result))
print(next(map_result))
print(next(map_result))
print(next(map_result))
```

```python
map_result = map(lambda_func, initial_list)
for element in map_result:
    print(element, end= ',')
```

```python
map_result = map(lambda_func, initial_list)
print(list(map_result))
```

```python
lambda_func = lambda i: i*2
initial_list= [1, 2, 3, 4, 5]
map_result = map(lambda_func, initial_list)
print(list(map_result))
```

```python
print(list(map(lambda i: i*2, [1, 2, 3, 4, 5]))) # same code as above just in one line
```

## Closures
#### Closure
- function defined inside another function that remembers the values of the outer function

```python
def greet(text):

    def print_greet():
        print(text)

    return print_greet

say_hello = greet('Hello')
say_hello()
```

```python
def make_multiply_closuer(x):

    def multiply(y):
        return x * y
    
    return multiply

multiply_5 = make_multiply_closuer(5)
multiply_12 = make_multiply_closuer(12)

print(multiply_5(10))
print(multiply_5(20))

print(multiply_12(10))
print(multiply_12(20))
```

## Text file reading basics

```python
try:
    stream = open('animals.txt')
    # your code goes here
    stream.close()  # method close
except Exception as e:
    print('An error occurred: ', e)
```

```python
try:
    stream = open('animals.txt')
    print(stream.read())
    stream.close()  # method close
except Exception as e:
    print('An error occurred: ', e)
```

```python
try:
    stream = open('animals.txt')
    print(stream.read(4)) # first 4 characters
    print(stream.read(2)) #second 2 characters
    print(stream.read(10000)) # rest of characters
    stream.close()  # method close
except Exception as e:
    print('An error occurred: ', e)
```

```python
try:
    stream = open('animals.txt')
    counter = 0
    character = stream.read(1)
    while character != '':
        counter +=1
        print(character, end='-')
        character = stream.read(1)
    print(stream.read())
    stream.close()  # method close
    print('\nNumber of characters:', counter)
except Exception as e:
    print('An error occurred: ', e)
```

## Reading binary files: read() vs readinto()
#### read() 
- method
- reads all the contents of the file into memory
- creates an object of a special class called bytes

--

- bytearray - mutable
- bytes - immutable

```python
try:
	bf = open('file.bin', 'rb')
	newly_created = bf.read()
	bf.close()
except Exception as e:
	print('An error occured:', e)
```

```python
# manual creation
data = bytearray(10)
try:
    bf = open('file.bin', 'rb')
    bf.readinto(data)
    bf.close()
except Exception as e:
    print('An error occured:', e)
```

## Stream errors

```python
try:
    stream = open('nonexistent.txt')
    stream.close()
except Exception as e:
    print('An error occured: ', e)
```

```python
try:
    stream = open('nonexistent.txt')
    stream.close()
except Exception as e:
    print(e.errno)
```

```python
from os import strerror

try:
    stream = open('nonexistent.txt')
    stream.close()
except Exception as e:
    print(strerror(e.errno))
```
