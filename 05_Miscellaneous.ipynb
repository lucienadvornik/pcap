## List comprehensions

```python
numbers = []
for i in range(1, 101):
    numbers.append(i)
```

```python
print('Numbers:', numbers)
```

```python
numbers = [i for i in range(1,101)]
print('Numbers:', numbers)
```

```python
numbers = [i for i in range(1,101) if i % 3 != 0]
print('Numbers:', numbers)
```

```python
numbers = [0 if i % 2 == 0 else 1 for i in range(100)]
print('Numbers:', numbers)
```

```python
table = [[i for i in range(1,6)] for j in range (5)]
print(table)
```

```python
def apply_func(elements, func):
    for element in elements:
        print(func(element))
```

```python
my_func = lambda x: x*x
apply_func([1,2,3,4,5], my_func)
```

```python
my_func = lambda x: 1/x
apply_func([1,2,3,4,5], my_func)
```

```python
my_func = lambda x: 0
apply_func([1,2,3,4,5], my_func)
```

```python
apply_func([1,2,3,4,5], lambda x: x*x)
```

# map() and filter() with lambdas

map()
- takes two arguments, function and sequence with some elements
- map() then applies the function to all the elements in the sequence

```python
lambda_func = lambda i: i*2
initial_list = [1, 2, 3, 4, 5]
```

```python
map_result = map(lambda_func, initial_list)
print(next(map_result))
print(next(map_result))
print(next(map_result))
print(next(map_result))
print(next(map_result))
print(next(map_result))
```

```python
map_result = map(lambda_func, initial_list)
for element in map_result:
    print(list(map_result))
```

## Stream errors

```python
try:
    stream = open('nonexistent.txt')
    stream.close()
except Exception as e:
    print('An error occured: ', e)
```

```python
try:
    stream = open('nonexistent.txt')
    stream.close()
except Exception as e:
    print(e.errno)
```

```python
from os import strerror

try:
    stream = open('nonexistent.txt')
    stream.close()
except Exception as e:
    print(strerror(e.errno))
```
