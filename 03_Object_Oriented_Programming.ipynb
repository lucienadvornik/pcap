# Object Oriented Programming

## Introduction to OOP

```python
# Procedural approach
def calculate_area(width, height):
    return width * height

width = int(input('What is the width? '))
height = int(input('What is the height? '))
area = calculate_area(width, height)
print('The area is', area)
```

```python
# Object approach
class Rectangle: 
    def __init__(self, width, height):
        self.width = width
        self.height = height

    def get_area(self):
        return self.width * self.height

a = int(input('What is the width? '))
b = int(input('What is the height? '))

rectangle = Rectangle(a,b)
print('The area is', rectangle.get_area())
```

## Classes and Objects

Classes
- can contain two types of entities
    - variables - called attributes

```python
# class User
# + username + city # set of properties
# + introduce() # set of method/s

# object
# ModernKnight 
# New York City
# + introduce()

# object
# RulerOfDarkness 
# London
# + introduce()
```

```python
class User:
    def __init__(self):     # constructor
        # whenever you create a user, give them two properties
        self.nickname = 'sampleNickname'
        self.city = 'sampleCity'

    def introduce(self):     # method
        print('Hello, I am', self.nickname, 'and I live in', self.city)

sample_user = User()
sample_user.introduce()
print(sample_user.nickname)
print(sample_user.city)
```

```python
class User:
    def __init__(self, nickname, city):     # constructor
        # whenever you create a user, give them two properties
        self.nickname = nickname
        self.city = city

    def introduce(self):     # method
        print('Hello, I am', self.nickname, 'and I live in', self.city)

sample_user = User('DarkKnight', 'Hell')
sample_user.introduce()
```

```python
first_user = User('Jane', 'London')
second_user = User('Eva', 'New York')
third_user = User('Martin', 'Boston')
```

```python
first_user.introduce()
second_user.introduce()
third_user.introduce()
```

## Encapulation and abstraction

```python
class Car():
    def __init__(self, model, colour, initial_speed = 0):
        self.model = model
        self.colour = colour
        if initial_speed <0:
            self.speed = 0
        else:
            self.speed = initial_speed
    
    def speed_up(self):
        self.speed +=5
    
    def speed_down(self):
        if self.speed < 5:
            self.speed = 0
        else:
            self.speed -=5
    
    def show_speed(self):
        print('Current speed:', self.speed)
```

```python
my_lovely_car = Car('T-rock', 'red')
my_lovely_car.speed_up()
my_lovely_car.speed_up()
my_lovely_car.show_speed()
```

```python
my_lovely_car = Car('T-rock', 'red')
my_lovely_car.speed_down()
my_lovely_car.speed_down()
my_lovely_car.speed_down()
my_lovely_car.show_speed()
```

```python
my_lovely_car = Car('T-rock', 'red')
my_lovely_car.speed = -10
my_lovely_car.show_speed()
```

### Refactoring

```python
class Car():
    def __init__(self, model, colour, initial_speed = 0):
        self.model = model
        self.colour = colour
        if initial_speed <0:
            self.__speed = 0 # private property - cannot be changed by my_lovely_car.speed = -10
        else:
            self.__speed = initial_speed
    
    def speed_up(self):
        self.__speed +=5
    
    def speed_down(self):
        if self.__speed < 5:
            self.__speed = 0
        else:
            self.__speed -=5
    
    def show_speed(self):
        print('Current speed:', self.__speed)
```

```python
my_lovely_car = Car('T-rock', 'red')
my_lovely_car.speed = -10
my_lovely_car.show_speed()
```

### OOP Principles
1. Encapsulation = objects should keep their state private and only expose a set of public function to the outside
2. Abstraction = objects should keep the details of how they work to themselves and only expose some high level operations to the outside
3. Inheritance
4. Polymorphism
