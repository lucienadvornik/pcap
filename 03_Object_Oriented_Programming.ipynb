# Object Oriented Programming

## Introduction to OOP

```python
# Procedural approach
def calculate_area(width, height):
    return width * height

width = int(input('What is the width? '))
height = int(input('What is the height? '))
area = calculate_area(width, height)
print('The area is', area)
```

```python
# Object approach
class Rectangle: 
    def __init__(self, width, height):
        self.width = width
        self.height = height

    def get_area(self):
        return self.width * self.height

a = int(input('What is the width? '))
b = int(input('What is the height? '))

rectangle = Rectangle(a,b)
print('The area is', rectangle.get_area())
```

## Classes and Objects

Classes
- can contain two types of entities
    - variables - called attributes
    - functions - called methods

```python
# class User
# + username + city # set of properties
# + introduce() # set of method/s

# object
# ModernKnight 
# New York City
# + introduce()

# object
# RulerOfDarkness 
# London
# + introduce()
```

```python
class User:
    def __init__(self):     # constructor
        # whenever you create a user, give them two properties
        self.nickname = 'sampleNickname'
        self.city = 'sampleCity'

    def introduce(self):     # method
        print('Hello, I am', self.nickname, 'and I live in', self.city)

sample_user = User()
sample_user.introduce()
print(sample_user.nickname)
print(sample_user.city)
```

```python
class User:
    def __init__(self, nickname, city):     # constructor
        # whenever you create a user, give them two properties
        self.nickname = nickname
        self.city = city

    def introduce(self):     # method
        print('Hello, I am', self.nickname, 'and I live in', self.city)

sample_user = User('DarkKnight', 'Hell')
sample_user.introduce()
```

```python
first_user = User('Jane', 'London')
second_user = User('Eva', 'New York')
third_user = User('Martin', 'Boston')
```

```python
first_user.introduce()
second_user.introduce()
third_user.introduce()
```
